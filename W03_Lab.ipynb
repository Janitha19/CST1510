### D6. A default value

1. Define `status_of(percent, warning_at=90)` that returns:
   - `"OVER LIMIT"` if `percent` is 100 or more
   - `"WARNING"` if `percent` is `warning_at` or more
   - `"OK"` otherwise
2. Print `status_of(95)`, then `status_of(85)`, then `status_of(85, 80)`.

**Expected output:** `WARNING`, `OK`, `WARNING`

````python
# D6
def status_of(percent, warning_at=90):
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"


print(status_of(95))
print(status_of(85))
print(status_of(85, 80))
````

### D8. `*args` and `**kwargs`

a) Define `total(*numbers)` that returns the sum of all the numbers passed in. Print `total(1, 2, 3)`,
   `total(10, 20, 30, 40)` and `total()`.

b) Define `configure(**settings)` that loops over `settings.items()` and prints each one as `key = value`.

**Expected output:**
```

```

````python
# D8
# a) Define total(*numbers) to sum arbitrary positional arguments
def total(*numbers):
    return sum(numbers)


print(total(1, 2, 3))
print(total(10, 20, 30, 40))
print(total())


# b) Define configure(**settings) to handle arbitrary keyword arguments
def configure(**settings):
    for key, value in settings.items():
        print(f"{key} = {value}")


configure(host="10.0.0.1", port=8080)
````

````python
# D9
# D9
def make_status():
    # Prediction: Will print 'inside: OK', then crash with a NameError on the outside print statement
    result = "OK"
    print("inside:", result)


make_status()
print(
    "outside:", result
)  # Explanation: 'result' is a local variable scoped only inside make_status(). Once the function finishes executing, its local variables are destroyed, so the global scope cannot see or access 'result'.
````

````python
# D10
# D10
def double(n):
    # Prediction: Will print 42 from the function call, then 21 for the global variable 'number'
    n = n * 2
    return n


number = 21
print(double(number))
print(
    number
)  # Explanation: Parameters like 'n' are local variables scoped entirely inside the function. Modifying 'n' inside double() does not affect the global variable 'number' outside the function.
````

### C1. Wire two functions together

Using `percent_of` from D3 and `status_of` from D6 **together**, print a status for each of these
three pairs. Do not write the percentage or status logic a second time.

| value | limit | Expected output |
|---|---|---|
| 87 | 100 | `OK` |
| 120 | 100 | `OVER LIMIT` |
| 45 | 100 | `OK` |

````python
# C1
# C1
def percent_of(part, total):
    """Return part as a percentage of total."""
    return (part / total) * 100

def status_of(percent, warning_at=90):
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"

# Wiring the two functions together for the three pairs (value, limit)
print(status_of(percent_of(87, 100)))
print(status_of(percent_of(120, 100)))
print(status_of(percent_of(45, 100)))
````

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python
# C2
# C2
def count_over(*values, limit=100):
    """Return how many of the values are over the limit."""
    count = 0
    for v in values:
        if v > limit:
            count += 1
    return count


# Test cases matching the table
print(count_over(87, 120, 45, 101))
print(count_over(87, 120, 45, 101, limit=50))
print(count_over())
````
