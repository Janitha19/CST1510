### D9. Loop over a dict

Using `record` from D7, write three loops:

a) print only the **keys** (`for key in record:`)

b) print only the **values** (`.values()`)

c) print both, as `key: value` (`.items()`)

**Expected output for c):**
```

```

````python
# D9
# D9
record = {"label": "R-001", "value": 90.0, "limit": 100.0, "status": "OK"}

# a) Print only the keys
for key in record:
    print(key)

print("-" * 20)  # Separator for clarity

# b) Print only the values using .values()
for value in record.values():
    print(value)

print("-" * 20)  # Separator for clarity

# c) Print both as key: value using .items()
for key, value in record.items():
    print(f"{key}: {value}")
````

### C1. Load records safely from a file

Write `load_records(path)` with a one-line docstring. It reads `label,value,limit` lines from a file
and returns a **list of dicts**. If the file does not exist, it prints a clear message and returns an
empty list instead of crashing.

Then load `records_sample.txt` and print each record's label and status, as in D16.

**Test it:**

| Call | Expected output |
|---|---|
| `load_records("records_sample.txt")` | the same three lines as D16 |
| `load_records("nope.txt")` | a clear message, then `[]` |

````python
# C1
def load_records(path):
    """Load records from a text file and return them as a list of dicts."""
    records = []
    try:
        with open(path, "r") as f:
            for line in f:
                label, value, limit = line.strip().split(",")
                record = {
                    "label": label,
                    "value": float(value),
                    "limit": float(limit)
                }
                records.append(record)
    except FileNotFoundError:
        print(f"Error: {path} not found.")
        return []
    return records

# 1. Load records_sample.txt and print each record's label and status (like D16)
records = load_records("records_sample.txt")
for r in records:
    if r["value"] > r["limit"]:
        status = "OVER LIMIT"
    else:
        status = "OK"
    print(f"{r['label']}: {status}")

print("-" * 30)

# 2. Test with a missing file (nope.txt)
print(load_records("nope.txt"))
````

### C2. Write the results to a file

Using your records from C1, write one line per record to a new file called `results.txt`,
in the form `label,status`. Then read `results.txt` back and print it.

**Expected output:**
```
R-001,OK
R-002,OVER LIMIT
R-003,OK
```

````python
# C2
# Using the records loaded from C1
records = load_records("records_sample.txt")

# 1. Write one line per record in the form label,status to "results.txt"
with open("results.txt", "w") as f:
    for r in records:
        if r["value"] > r["limit"]:
            status = "OVER LIMIT"
        else:
            status = "OK"
        f.write(f"{r['label']},{status}\n")

# 2. Read "results.txt" back and print it
with open("results.txt", "r") as f:
    print(f.read(), end="")
````
