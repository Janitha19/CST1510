### D6. `self` is whichever object made the call

1. Copy your class from D5 and add a method `status(self)` that returns `"OVER LIMIT"` (100% or more),
   `"WARNING"` (90% or more) or `"OK"`, worked out from `self.value` and `self.limit`.
2. Create three records and print each one's `.status()`:

| Record | Expected output |
|---|---|
| `Record("R-001", 87.0, 100.0)` | `OK` |
| `Record("R-002", 120.0, 100.0)` | `OVER LIMIT` |
| `Record("R-003", 95.0, 100.0)` | `WARNING` |

````python
# D6
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        return f"{self.label}: {self.value}/{self.limit}"

    # 1. Add the status method with conditional thresholds
    def status(self):
        if self.value >= self.limit:
            return "OVER LIMIT"
        elif self.value >= 0.9 * self.limit:  # 90% or more
            return "WARNING"
        else:
            return "OK"

# 2. Create three records and print each one's .status()
r1 = Record("R-001", 87.0, 100.0)
r2 = Record("R-002", 120.0, 100.0)
r3 = Record("R-003", 95.0, 100.0)

print(r1.status())
print(r2.status())
print(r3.status())
````

### D8. A method can change its own object

### D9. A list of objects

1. Using your `Record` class from D7, create a list of three records:
   `R-001` (87.0), `R-002` (120.0) and `R-003` (45.0), each with a limit of 100.0.
2. Loop over the list and print each record's label and status.

**Expected output:**

````python
# D9
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        return f"{self.label}: {self.value}/{self.limit}"

    def percent(self):
        return (self.value / self.limit) * 100

    def status(self):
        pct = self.percent()
        if pct >= 100.0:
            return "OVER LIMIT"
        elif pct >= 90.0:
            return "WARNING"
        else:
            return "OK"

# 1. Create a list of three records
records = [
    Record("R-001", 87.0, 100.0),
    Record("R-002", 120.0, 100.0),
    Record("R-003", 45.0, 100.0)
]

# 2. Loop over the list and print each record's label and status
for r in records:
    print(r.label, r.status())
````

### D11. Load objects from a file

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

1. Write `load_records(path)` that reads the file and returns a list of `Record` objects
   (convert `value` and `limit` with `float()`).
2. Loop over the list and print each record's label and status.

**Expected output:** the same three lines as D9.

````python
# D11
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        return f"{self.label}: {self.value}/{self.limit}"

    def percent(self):
        return (self.value / self.limit) * 100

    def status(self):
        pct = self.percent()
        if pct >= 100.0:
            return "OVER LIMIT"
        elif pct >= 90.0:
            return "WARNING"
        else:
            return "OK"

# 1. Write load_records(path) returning a list of Record objects
def load_records(path):
    records = []
    with open(path, "r") as f:
        for line in f:
            label, value, limit = line.strip().split(",")
            # Instantiate a Record object with float conversions for numeric fields
            record_obj = Record(label, float(value), float(limit))
            records.append(record_obj)
    return records

# 2. Loop over the list and print each record's label and status
records = load_records("records_sample.txt")
for r in records:
    print(r.label, r.status())
````

### C1. Count the records over the limit

Add a method `is_over(self)` to `Record` that returns `True` or `False`. Load `records_sample.txt` with
`load_records` from D11, use `is_over()` to count how many records are over their limit, and print a
one-line summary.

**Expected output:** `1 of 3 records are OVER LIMIT`

````python
# C1
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        return f"{self.label}: {self.value}/{self.limit}"

    def percent(self):
        return (self.value / self.limit) * 100

    def status(self):
        pct = self.percent()
        if pct >= 100.0:
            return "OVER LIMIT"
        elif pct >= 90.0:
            return "WARNING"
        else:
            return "OK"

    # 1. Add is_over(self) method returning True or False
    def is_over(self):
        return self.value > self.limit

def load_records(path):
    records = []
    with open(path, "r") as f:
        for line in f:
            label, value, limit = line.strip().split(",")
            records.append(Record(label, float(value), float(limit)))
    return records

# 2. Load records_sample.txt and use is_over() to count records over their limit
records = load_records("records_sample.txt")
over_count = sum(1 for r in records if r.is_over())
total_count = len(records)

print(f"{over_count} of {total_count} records are OVER LIMIT")
````

### C2. An account that locks

Write a class `Account` with a `username` and a count of failed logins that starts at 0.
It has two methods:

- `fail(self)`: adds 1 to the failed logins
- `locked(self)`: returns `True` if there have been 3 or more failed logins

**Test it:**

| Code | Expected output |
|---|---|
| an account for `"anna"` with `.fail()` called 3 times, then `print(anna.locked())` | `True` |
| an account for `"ben"` with `.fail()` called once, then `print(ben.locked())` | `False` |

````python
# C2
class Account:
    def __init__(self, username):
        self.username = username
        self.failed_logins = 0

    def fail(self):
        """Adds 1 to the failed logins counter."""
        self.failed_logins += 1

    def locked(self):
        """Returns True if there have been 3 or more failed logins."""
        return self.failed_logins >= 3

# Test case 1: Account for "anna" with .fail() called 3 times
anna = Account("anna")
anna.fail()
anna.fail()
anna.fail()
print(anna.locked())  # Expected: True

# Test case 2: Account for "ben" with .fail() called once
ben = Account("ben")
ben.fail()
print(ben.locked())   # Expected: False
````
